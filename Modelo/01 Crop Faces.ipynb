```python
import pandas as pd
import numpy as np
from PIL import Image
from tqdm import tqdm
from io import BytesIO
from deepface import DeepFace
import cv2
import os
```

```python
df = pd.read_parquet('./UTKFACE_dataset.parquet')
df.head()
```

```python
print(f"Edad Minima: {df['age'].min()}")
print(f"Edad Maxima: {df['age'].max()}")
```

```python
def load_image(image_path):
    
    img_pillow = Image.open(image_path)
    img_array = np.array(img_pillow)
    return img_array
```

```python
images_cropped = []
images_problem = []
for user in tqdm(df.to_dict(orient = 'records')):
    
    image_path = user['image_path']
    image_name = os.path.basename(image_path).split('.')[0]
    if os.path.exists(f'./UTKFACE_cropped/{image_name}'):
        images_cropped.append(image_path)
        continue
    
    try:
        image_array = load_image(image_path)
        
        img_objs = DeepFace.extract_faces(
            img_path = image_array,
            detector_backend = 'retinaface', 
            align = True, 
            expand_percentage = 10,
            enforce_detection = False
        )
        
        area = img_objs[0]['facial_area']
        x, y, w, h = area['x'], area['y'], area['w'], area['h']
        face_crop = image_array[y:y+h, x:x+w]
    
        face_img = Image.fromarray(face_crop)
        face_img.save(f'./UTKFACE_cropped/{image_name}.jpg')
        images_cropped.append(image_path)
        
    except:
        print(f'Problems with: {image_name}')
        images_problem.append(image_path)
        continue
        
```

```python
df = df[df['image_path'].isin(images_cropped)]
print(len(df))
```

```python
df.to_parquet('./UTKFACE_dataset.parquet')
```
